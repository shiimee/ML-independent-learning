## Load libraries

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Line graphs

```python
# Basic line graph
x = [0,1,2,3,4]
y = [0,2,4,6,8]

# Resize graph
plt.figure(figsize=(5,3), dpi=300)

#--- Line 1 ---#
# Keyword arguement notation
# plt.plot(x,y, label='2x', color='red', linewidth=2, marker='.', markersize=10, markeredgecolor='blue', linestyle='--')

# Shorthand notation
# fmt = '[color][marker][linestyle]'
plt.plot(x,y, 'b^-', label='2x')

#--- Line 2 ---#

# Selecting interval we want to plot points at
x2 = np.arange(0,4.5,0.5)

# Plot part of the graph as a line
plt.plot(x2[:6], x2[:6]**2, 'r-', label='x^2') 

# Plot remainder of graph as dashed lines
plt.plot(x2[5:], x2[5:]**2, 'r--')

# Add title (specify font params with fontdict, fontdict can be used with any text based code)
plt.title('First Graph', fontdict={'fontname': 'Comic Sans MS', 'fontsize': 20} )

# X and Y labels
plt.xlabel('x-axis')
plt.ylabel('y-axis')

# X, Y tickmarks (scale of graph)
plt.xticks([0,1,2,3,4])
plt.yticks([0,2,4,6,8,10])

# legends
plt.legend()

# Saving the plot in current directory (dpi 300 is high reso standard)
# plt.savefig('mygraph.png', dpi=300)

# Show plot
plt.show()
```

## Bar charts

## Real World Examples

gas prices data

```python
# Reading the data
gas = pd.read_csv('gas_prices.csv')
print('read file ok')

gas
```

```python
# Line graphs

plt.figure(figsize=(8,5), dpi=300)

plt.title('Gas Prices Over Time (USD)', fontdict={})

for country in gas.columns[1:]:
    plt.plot(gas['Year'], gas[country], marker='.', label=country)

# plt.plot(gas['Year'], gas['USA'], 'b.-', label='USA') # note: positional arguements come before keyword arguements
# plt.plot(gas['Year'], gas['Canada'], 'r.-', label='Canada')
# plt.plot(gas['Year'], gas['South Korea'], 'g.-', label='South Korea')
# plt.plot(gas['Year'], gas['Australia'], 'y.-', label='Australia')

plt.xticks(gas['Year'][::3].tolist()+[2011]) # [::3] all years from start to end, step of 3

plt.xlabel('Year')
plt.ylabel('US Dollars')

plt.legend()

plt.show()
```

fifa data 

```python
fifa = pd.read_csv('fifa_data.csv')
fifa.head()
```

```python
# histogram
bins = np.arange(40,101,10)
print(bins)

plt.hist(fifa['Overall'], bins=bins, color='#56bab2') # color use color picker

plt.xticks(bins)

plt.ylabel('Number of players')
plt.xlabel('Skill level')
plt.title('Distribution of player skills in FIFA 2018')


plt.show()
```

```python
# pie chart 1
left = fifa.loc[fifa['Preferred Foot'] == 'Left'].count()[0]
right = fifa.loc[fifa['Preferred Foot'] == 'Right'].count()[0]

labels = ['Left', 'Right']
colors = ['#abcdef', '#aabbcc']

plt.pie([left, right], labels=labels, colors=colors, autopct='%.2f %%')

plt.title('Foot Preferrence of FIFA Players')

plt.show()
```

```python
# pie chart 2
fifa['Weight'] = [int(x.strip('lbs')) if type(x) == str else x for x in fifa['Weight']]
fifa.Weight[0]

plt.style.use('ggplot')

light = fifa.loc[fifa['Weight'] < 125].count()[0]
light_med = fifa.loc[(fifa['Weight'] >= 125) & (fifa['Weight'] < 150)].count()[0]
med = fifa.loc[(fifa['Weight'] >= 150) & (fifa['Weight'] < 175)].count()[0]
med_heavy = fifa.loc[(fifa['Weight'] >= 175) & (fifa['Weight'] < 200)].count()[0]
heavy = fifa.loc[fifa['Weight'] >= 200].count()[0]

weights = [light, light_med, med, med_heavy, heavy]
labels = ['Under 125', '125-150', '150-175', '175-200', 'Over 200']
explode = (.4,.1,0,0,.4)

plt.title('Weight Distribution of FIFA Players (in lbs)')

plt.pie(weights, labels=labels, autopct='%.2f %%', pctdistance=0.8, explode=explode)

plt.show()
```

```python
# box and whisker
plt.style.use('default')

plt.figure(figsize=(5,8))

barcelona = fifa.loc[fifa.Club == 'FC Barcelona']['Overall']
madrid = fifa.loc[fifa.Club == 'Real Madrid']['Overall']
revs = fifa.loc[fifa.Club == 'New England Revolution']['Overall']

labels = ['FC Barcelona', 'Real Madrid', 'NE Revolution']

boxes = plt.boxplot([barcelona, madrid, revs], labels=labels, showmeans=True, patch_artist=True, medianprops={'linewidth':2})

for box in boxes['boxes']:
    # set edge color
    box.set(color='#4286f4', linewidth=2)

    # fill color
    box.set(facecolor='#e0e0e0')

plt.title('Professional Soccer Team Comparison')
plt.ylabel('FIFA Overall Rating')

plt.show()
```

## Subplots 

Figure = Entire canvas 🖼️

Ax = A single plot (subplots) 📈

```python
# dummy data
x = np.arange(1,6)

plt.style.use('ggplot')

# create 2x2 subplot
fig, ax = plt.subplots(2,2)

# first plot (top left)
ax[0,0].plot(x, x*2, color='red')
ax[0,0].set_title("x*2")

# second plot (top right)
ax[0,1].plot(x, x**2, color='blue')
ax[0,1].set_title('x**2')

# third plot (bottom left)
ax[1,0].plot(x, x**3, color='green')
ax[1,0].set_title('x**3')

# fourth plot (bottom right)
ax[1,1].plot(x, x**4, color='purple')
ax[1,1].set_title('x**4')


# prevent overlap
plt.tight_layout()

plt.show()
```
